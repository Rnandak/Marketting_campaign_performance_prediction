```python
from sklearn.linear_model import LinearRegression
```

```python
from sklearn.preprocessing import StandardScaler,MinMaxScaler
```

```python
f = consolidatedlinearencoded[["Conversions","ROI","Engagement_Score","Leads","Clicks","Impressions"]]
```

```python
t = consolidatedlinearencoded[["Revenue"]]
```

```python
from sklearn.pipeline import Pipeline
```

```python
from sklearn.model_selection import train_test_split
```

```python
f_train,f_test,t_train,t_test = train_test_split(f,t,test_size=0.2,random_state=4)
```

```python
lrmodel = Pipeline([
    ("scaling",MinMaxScaler()),
    ("linear",LinearRegression())
])
```

```python
lrmodel.fit(f_train,t_train)
```

```python
t_pred = lrmodel.predict(f_test)
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(t_pred,t_test)
```
